```python
# 1.Smart parking lot management system

# function 
def parking_system(cap, logs):
    parked = 0  

    for log in logs:
        log = log.upper()
        if log == "IN":
            parked += 1
        elif log == "OUT":
            if parked > 0:
                parked -= 1
                
        # if capacity exceed
        if parked > cap:
            print("Alert: Parking Full!")
            
    # parking availability
    print("Currently Parked Vehicles:", parked)
    if parked >= cap:
        print("Parking Status: Full")
    else:
        print("Parking Status: Available")


# user input
capacity = int(input("Enter parking capacity: "))
logs_input = input("Enter vehicle logs separated by space (IN/OUT): ")

vehicle_logs = logs_input.split()

# function call
parking_system(capacity, vehicle_logs)
```

```python
# 2.online food delivery time estimator

# function 
def delivery_time(distance, traffic, weather):
    base_time = distance * 5  # 5 min per km

    # Traffic delays
    if traffic.lower() == "high":
        base_time += 15
    elif traffic.lower() == "medium":
        base_time += 5

    # weather delays
    if weather.lower() == "rainy":
        base_time += 5   # reduced from 10 to 5
    elif weather.lower() == "stormy":
        base_time += 20

    print("Estimated Delivery Time:", base_time, "minutes")


# user input
distance = int(input("Distance (km): "))
traffic = input("Traffic Level (Low/Medium/High): ")
weather = input("Weather (Clear/Rainy/Stormy): ")

# function call
delivery_time(distance, traffic, weather)
```

```python
# 3. movie theatre seat occupancy analyzer

# function
def seat_analyzer(total_seats, booked_seats):
    booked = len(booked_seats)
    percentage = int((booked / total_seats) * 100)   

    print("Occupancy:", percentage, "%")

    if percentage == 100:
        print("Show Status: Housefull")
        print("Suggestion: Open new show")
    elif percentage >= 75:
        print("Show Status: Almost Full")
        print("Suggestion: No need for new show")
    else:
        print("Show Status: Seats Available")
        print("Suggestion: No need for new show")


# user input
total_seats = int(input("Enter total seats: "))
booked_count = int(input("Enter number of booked seats: "))

# Creating list with booked seats
booked_seats = [1] * booked_count

# function call
seat_analyzer(total_seats, booked_seats)
```

```python
# 4.cloud server load classification system

# function
def server_load(cpu_readings):
    total = 0
    
    # add readings
    for i in cpu_readings:
        total = total + i
    
    # calculate average
    average = int(total / len(cpu_readings))
    
    print("Average CPU Load:", average, "%")
    
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")


# user input
readings = input("Enter CPU readings separated by space: ")

# list
cpu_list = []
for value in readings.split():
    cpu_list.append(float(value))
    
# function call
server_load(cpu_list)
```

```python
# 5. smart classroom resource usage monitor

# function
def resource_monitor(usage):
    overused = []

    if usage["Projector"] > 6:  
        overused.append("Projector")
    if usage["AC"] > 8:
        overused.append("AC")
    if usage["Lights"] > 6:
        overused.append("Lights")

    if len(overused) == 0:
        print("Overused Resources: None")
        print("Energy Alert: No")
    else:
        print("Overused Resources:", end=" ")
        for resource in overused:
            print(resource, end=" ")
        print()  
        print("Energy Alert: Yes")


# user input
projector_hours = int(input("Enter Projector usage hours: "))
ac_hours = int(input("Enter AC usage hours: "))
lights_hours = int(input("Enter Lights usage hours: "))

# dictionary
resource_usage = {
    "Projector": projector_hours,
    "AC": ac_hours,
    "Lights": lights_hours
}

# function call
resource_monitor(resource_usage)
```

```python
# 6.  online event registration capacity controller

# function
def manage_registrations(capacity, registrations):

    confirmed = min(registrations, capacity)
    waitlisted = registrations - confirmed

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    
    # status
    if waitlisted > 0:
        print("Registration Status: Closed")
    else:
        print("Registration Status: Open")


# user input
event_capacity = int(input("Enter event capacity: "))
total_registrations = int(input("Enter number of registrations: "))

# function call
manage_registrations(event_capacity, total_registrations)
```
