```python
# 1.Mobile recharge validation system

# function for vlidation of recharge
def recharge_validation(amount):
    valid_plans = [199, 299, 399, 599]
    
    if amount < 50:
        return "Recharge amount must be at least ₹50."
    elif amount not in valid_plans:
        return "Invalid plan selected. plan must be "
    else:
        return "Recharge successful!"

# loop for retry
while True:
    try:
        amount = int(input("Enter recharge amount:"))
        result = recharge_validation(amount)
        print(result)
        
        if result == "Recharge successful!":
            break
        else:
            print("Please try again.\n")
            
    except ValueError:
        print("Invalid input. Please enter a numeric value.\n")
```

```python
# 2. Inventory Reorder Alert System

# function to check stock
def check_stock(products):
    for item in products:
        if products[item] < 15:
            print(item, "- Reorder Alert")
        else:
            print(item, "- Stock OK")

# Dictionary of products
inventory = {
    "Rice": 20,
    "Sugar": 8,
    "Oil": 12,
    "Soap": 25
}

check_stock(inventory)
```

```python
# 3. Student Result Processing System

# Function to calculate result
def student_results(marks):
    total = 0 

    # Loop through each mark and add to total
    for mark in marks:
        total += mark

    # Calculate average
    average = total / len(marks)

    # Pass or Fail
    if average >= 50:
        return "Pass"
    else:
        return "Fail"

# user input
marks_input = input("Enter student marks: ").split()

# Convert input to numbers
marks = []
for mark in marks_input:
    marks.append(float(mark))  

# Call the function
result = student_results(marks)

# Print the result
print("Result:", result)
```

```python
# 4. Cab Fare Estimator with Retry Option

# Function to calculate fare
def calculate_fare(distance, peak_hour):
    base_fare = 50  
    per_km = 12     

    # Calculate fare without peak hrs
    fare = base_fare + (per_km * distance)

    # 25% extra for peak hour
    if peak_hour.lower() == 'yes':
        fare += fare * 0.25

    return fare

# loop for retry
while True:
    # user input
    distance = float(input("Enter distance traveled (in km): "))
    peak_hour = input("Is it peak hour? (Yes/No): ")

    # Calculate fare
    total_fare = calculate_fare(distance, peak_hour)
    print("Total Fare: ₹", total_fare)

    #  if user wants to calculate again
    retry = input("Do you want to calculate fare again? (Yes/No): ")
    if retry.lower() != 'yes':
        print("Thank you for using the Cab Fare Calculator!")
        break
```

```python
# 5. Employee Attendance Eligibility Checker

# Function to check attendance eligibility
def check_attendance(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    # Count number of present days
    for day in attendance_list:
        if day.upper() == 'P':
            present_days += 1

    # Calculate attendance percentage
    percentage = (present_days / total_days) * 100

    # Checking eligibility
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# user input
attendance_input = input("Enter attendance record (P for present, A for absent):")

# Call the function
result = check_attendance(attendance_input)

# Print the result
print("Attendance Status:", result)
```

```python
# 6.Password Strength Checker

# Function to check password strength
def password_strength(password):
    special_chars = "@#$"
    has_digit = False
    has_special = False

    # Check each character in password
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True

    # Check all conditions
    if len(password) >= 8 and has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

# user input
password = input("Enter your password: ")

# Call the function
result = password_strength(password)

# Print the result
print(result)
```
