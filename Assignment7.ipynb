```python
# 1. smart light controller
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f"{self.name} is ON")

    def turn_off(self):
        self.is_on = False
        print(f"{self.name} is OFF")

    def display_status(self):
        status = "ON" if self.is_on else "OFF"
        print(f"{self.name} is {status}")


# create an object
smartlight = SmartLight("Bedroom Light")
smartlight.turn_on()   
```

```python
# 2. employee id card system
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department

    def display_id_details(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# create an object
emp = Employee("Rahul", "EMP102", "IT")
emp.display_id_details()
```

```python
# 3. mobile contact record
class Contact:
    
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def display_contact_details(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# create an object
contact = Contact("Anita", "9876543210")
contact.display_contact_details()
```

```python
# 4. product price tag generator
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")


# create an object
product = Product("Headphones", 2499)
product.display_price_tag()
```

```python
# 5. movie rating display system
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def display_movie_details(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")


# create an object
mv = Movie("Inception", 4.8)
mv.display_movie_details()
```

```python
# 6. delivery address manager
class Delivery:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address

    def display_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# create an object
delivery = Delivery("Suman", "Hyderabad")
delivery.display_delivery_details()
```
