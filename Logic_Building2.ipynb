```python
# 1 Unique Words in a Sentence

# Given
sentence = "python is easy and python is powerful"

# Split the sentence into individual words
words = sentence.split()

# Create an empty list to store unique words
unique = []

# Loop through each word in the list
for w in words:
    # Check if the word is not already in unique list
    if w not in unique:
        # Add the word to the unique list
        unique.append(w)

print("Unique words count:", len(unique))

print("Unique words: {'" + "', '".join(unique) + "'}")
```

```python
# 2 Highest Salary from Employee Data

# Dictionary of employees and their salary
employees = {"Ravi": 75000, "Anita": 68000, "Kiran": 72000}

# Assume first employee has highest salary
emp_name = ""
highest_salary = 0

# Loop through dictionary
for name in employees:
    # Check if current salary is greater than highest_salary
    if employees[name] > highest_salary:
        highest_salary = employees[name]
        emp_name = name

print("Highest Salary:", emp_name, "-", highest_salary)
```

```python
# 3 Find Maximum and Minimum Values

# List of numbers
numbers = [45, 22, 89, 10, 66]

# Initialize maximum and minimum with the first element
maximum = numbers[0]
minimum = numbers[0]

# Loop through each number in the list
for num in numbers:
    # Update maximum if current number is greater
    if num > maximum:
        maximum = num
    # Update minimum if current number is smaller
    if num < minimum:
        minimum = num

print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

```python
# 4 Count Products Above a Price Threshold

# List of product prices
prices = [450, 1200, 899, 1500, 300]
count = 0 
# Check each price
for price in prices: 
    if price > 1000:
        count += 1 
        
print("Products above 1000:", count) 
```

```python
# 5 Calculate Attendance Percentage

# Attendance list
attendance = ["P", "P", "A", "P", "P"]

# Count number of Present days
present = attendance.count("P")

# Total number of days
total_days = len(attendance)

# Calculate percentage
percentage = (present / total_days) * 100

print("Attendance:", percentage)
```

```python
# 6 Remove Duplicate Phone Numbers

# List with duplicate phone numbers
phone_numbers = [9876543210, 9123456789, 9876543210]

# Convert list to set to remove duplicates
unique_numbers = set(phone_numbers)

print("Unique phone numbers:", unique_numbers)
```

```python
# 7 Count Character Frequency

# Given 
text = "ppython"

# Empty dictionary to store frequency
frequency = {}

# Loop through each character in the string
for char in text:
    if char in frequency:
        frequency[char] += 1  # Increment count if already in dictionary
    else:
        frequency[char] = 1   # Initialize count as 1 if first occurrence
        
print(frequency)
```

```python
# 9 Check if a Key Exists in a Dictionary

# Employee dictionary
employees = {"Ravi": 75000, "Anita": 68000}

# Check if key exists
if "Ravi" in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```

```python
# 10 Calculate Average Marks

# List of marks
marks = [70, 75, 80, 65]

# Calculate total marks
total = sum(marks)

# Calculate average
average = total / len(marks)

print("Average Marks:", average)
```
