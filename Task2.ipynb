```python
# 2. Pass / Fail Analyzer

marks = [45, 78, 90, 33, 60]

total_pass = 0
total_fail = 0

# Loop for marks
for mark in marks:
    if mark >= 50:
        total_pass += 1
    else:
        total_fail += 1

# Print result
print("\nPass / Fail Analysis")
print("Total Students Passed:", total_pass)
print("Total Students Failed:", total_fail)
```
