```python
# 1.Social Media – Post Engagement Analyzer
# Function to analyze social media post engagement
def analyze_post(likes):
    total_likes = 0  # Initialize total likes
    
    #  sum all likes
    for like in likes:
        total_likes += like
    
    # post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    # Print the results
    print("Total Likes:", total_likes)
    print("Post Status:", status)


# user input for the number of likes entries
n = int(input("Enter the number of likes entries: "))

# empty list to store likes
likes = []

# user input from the user
for i in range(n):
    like = int(input(f"Enter likes for entry {i+1}: "))
    likes.append(like)

# function call
analyze_post(likes)
```

```python
# 2. Healthcare – Medicine Stock Alert System
# Function to monitor medicine stock
def check_medicine_stock(stock):
    # Determine the status based on stock level
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"
    
    # Print the results
    print("Medicine Stock:", stock)
    print("Status:", status)


# user input 
stock = int(input("Enter the medicine stock level: "))

# function call
check_medicine_stock(stock)
```

```python
# 3. Agriculture – Rainfall Adequacy Checker
# Function to analyze seasonal rainfall
def check_rainfall(rainfall_data, required_level):
    total_rainfall = 0  # Initialize total rainfall6
    
    
    # rainfall data to calculate total
    for rain in rainfall_data:
        total_rainfall += rain
    
    # Calculate average rainfall
    average_rainfall = total_rainfall / len(rainfall_data)
    
    # rainfall status
    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    # Print results
    print("Average Rainfall:", round(average_rainfall))
    print("Rainfall Status:", status)


# user input for number of months in the season
n = int(input("Enter number of months in the season: "))

#  rainfall input for each month
rainfall_data = []
for i in range(n):
    rain = float(input(f"Enter rainfall for month {i+1}: "))
    rainfall_data.append(rain)

# required average rainfall level input
required_level = float(input("Enter required average rainfall: "))

# function call
check_rainfall(rainfall_data, required_level)
```

```python
# 4. Social Media – Duplicate Account Detection
# Function to detect duplicate usernames
def check_duplicates(usernames):
    # Convert the list to a set to remove duplicates
    unique_usernames = set(usernames)
    
    # Compare lengths to see if duplicates exist
    if len(unique_usernames) < len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


#  user input for number of usernames
n = int(input("Enter the number of usernames: "))

#  usernames input from the user
usernames = []
for i in range(n):
    username = input(f"Enter username {i+1}: ")
    usernames.append(username)

# function call
check_duplicates(usernames)
```

```python
# 6.  Agriculture – Premium Crop Price Filter
# Function to filter premium crops
def premium_crops(prices):
    result = []
    for price in prices:
        if price > 2000: 
            result.append(price)
    print("Premium Crops:", result)


# user nput
n = int(input("Enter number of crops: "))
prices = []
for i in range(n):
    prices.append(int(input(f"Enter price for crop {i+1}: ")))

# function call
premium_crops(prices)
```

```python
# 7.  System Monitoring – Application Health Checker
# Function to check application health
def check_app_health(errors):
    # system status based on error count
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"
    
    # Print results
    print("Error Count:", errors)
    print("System Status:", status)


# user input 
errors = int(input("Enter number of errors: "))

# function call
check_app_health(errors)
```

```python
# 8. Banking – Daily Transaction Limit Checker
# Function to check daily transaction limit
def check_transaction(amount):
    daily_limit = 50000
    print(f"Transaction Amount: {amount}")
    
    if amount <= daily_limit:
        print("Transaction Status: Approved")
    else:
        print("Transaction Status: Rejected")

#  user input
amount = float(input("Enter transaction amount: ₹"))
# function call
check_transaction(amount)
```

```python
# 9. E-Learning – Student Attendance Eligibility System
# Function to calculate attendance percentage and eligibility
def check_attendance(attendance):
    percentage = (sum(attendance) / len(attendance)) * 100
    print(f"Attendance Percentage: {percentage:.1f}")
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")

#  empty list to store attendance
attendance = []

# user input
num_classes = int(input("Enter total number of classes: "))
for i in range(num_classes):
    status = input(f"Was the student present in class {i+1}? (yes/no): ").lower()
    attendance.append(1 if status == 'yes' else 0)

# function call
check_attendance(attendance)
```

```python
# 10.  Smart Electricity Bill Analyzer
# Function to calculate bill and usage status
def calculate_bill(units):
    # Bill calculation 
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = 100 * 3 + (units - 100) * 5
    else:
        bill = 100 * 3 + 100 * 5 + (units - 200) * 7

    
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage

# user input
units = float(input("Enter electricity units consumed: "))

# Calculate and display results
bill, usage_status = calculate_bill(units)
print(f"Total Bill: ₹{bill}")
print(f"Usage Classification: {usage_status}")
```
